AI Telecom Challenge 2026 NBO

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```
df_clientes = pd.read_csv('../data/dataset_clientes.csv')
df_ofertas = pd.read_csv('../data/catalogo_ofertas_entrega.csv')
df_campanias = pd.read_csv('../data/historial_campanias.csv')
```

## 1. Movistar Total Eligibility

```
df_clientes['elegible_mt'].value_counts(normalize=True).plot(kind='bar')
plt.title('MT Eligibility')
plt.show()
```

## 2. Acceptance Rates

```
df_valid = df_campanias[df_campanias['resultado'] != 'pendiente'].copy()
df_valid['aceptada'] = (df_valid['resultado'] == 'aceptada').astype(int)
df_valid.groupby('canal')['aceptada'].mean().sort_values().plot(kind='barh')
plt.title('Acceptance by Channel')
plt.show()
```
